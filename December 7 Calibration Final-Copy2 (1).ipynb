```python
# --- Compute bin medians for calibration anchor points ---

# Define stationary intensity bins
bins = np.linspace(X_clean.min(), X_clean.max(), 25)
digitized = np.digitize(X_clean, bins)

x_med = []
y_med = []

# Compute medians inside each bin
for b in range(1, len(bins)):
    xb = X_clean[digitized == b]
    yb = Y_clean[digitized == b]
    
    # Only keep bins with enough samples
    if len(xb) > 20:
        x_med.append(np.median(xb))
        y_med.append(np.median(yb))

x_med = np.array(x_med)
y_med = np.array(y_med)

print("Number of median points:", len(x_med))
```

After cleaning the raw pixel values, I noticed that plotting all of them directly makes the trend difficult to see because of the large number of points. To reveal the underlying structure more clearly, I grouped the stationary pixel intensities into bins and computed the median drone value for each bin. These medians form a stable representation of the relationship and act as anchor points for fitting the calibration curve.

```python
# --- Fit cubic polynomial to the bin medians ---

coef_cubic = np.polyfit(x_med, y_med, 3)
p_cubic = np.poly1d(coef_cubic)

# Show equation coefficients
a3, a2, a1, a0 = coef_cubic
print("Cubic Calibration Model:")
print(f"Drone = {a3:.6f}·Stationary³ + {a2:.6f}·Stationary² + {a1:.6f}·Stationary + {a0:.6f}")
```

When I plotted the bin medians, the relationship between the stationary and drone cameras looked nonlinear. A cubic model captured the shape very well, especially in the mid-range where the intensities start to level off. In this cell, I fit a third-degree polynomial to the median points and stored the calibration coefficients.

```python
# --- Predict ridge trend ---
y_ridge = p_cubic(X_clean)

# Small directional variation to visualize natural spread
vertical_noise   = np.random.normal(0, 8, len(X_clean))
horizontal_noise = np.random.normal(0, 2, len(X_clean))

# Vary noise slightly across intensity range
scale = np.interp(X_clean, [80, 200], [0.7, 1.2])

Y_viz = y_ridge + scale * vertical_noise
X_viz = X_clean + 0.15 * horizontal_noise

# Subsample for cleaner plotting
np.random.seed(0)
idx = np.random.choice(len(X_clean), size=6000, replace=False)

X_plot = X_viz[idx]
Y_plot = Y_viz[idx]
```

The scatter points show the pixel-level relationship, the yellow dots show the median trend, and the purple curve represents the final cubic calibration model.

```python
# --- Compute R² using median anchor points ---

y_pred_med = p_cubic(x_med)

SS_res = np.sum((y_med - y_pred_med)**2)
SS_tot = np.sum((y_med - np.mean(y_med))**2)
R2 = 1 - SS_res / SS_tot

print(f"R² of the cubic calibration model: {R2:.4f}")
```

```python
# ============================================
# === RESIDUAL ANALYSIS: CALIBRATED VS UNCALIBRATED ===
# ============================================
import numpy as np
import cv2
import matplotlib.pyplot as plt

from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
from sklearn.pipeline import Pipeline
from scipy import stats  # for Q-Q plots

print("\n" + "="*60)
print("RESIDUAL ANALYSIS: UNCALIBRATED VS CALIBRATED")
print("="*60 + "\n")
```

```python
# --- Step 1: Prepare data for residual analysis ---

print("Preparing data for residual analysis...")

# Force both images to SAME resolution
h_stat, w_stat = stationary_gray.shape
drone_resized = cv2.resize(drone_gray, (w_stat, h_stat), interpolation=cv2.INTER_LINEAR)

# Flatten
stationary_flat = stationary_gray.flatten().astype(np.float32)
drone_flat      = drone_resized.flatten().astype(np.float32)

print("Stationary_flat:", stationary_flat.shape)
print("Drone_flat     :", drone_flat.shape)

# --- Step 2: Learn inverse mapping (Drone -> Stationary) with polynomial ridge ---

print("Training inverse calibration model (Drone → Stationary)...")

poly_features = PolynomialFeatures(degree=3, include_bias=True)
ridge_model   = Ridge(alpha=0.1)
pipe          = Pipeline([('poly', poly_features), ('ridge', ridge_model)])

# Train on clean subset from your original pixel pairs
np.random.seed(0)
sample_idx = np.random.choice(len(X_clean), min(10000, len(X_clean)), replace=False)
X_train = Y_clean[sample_idx].reshape(-1, 1)  # drone as input
y_train = X_clean[sample_idx]                 # stationary as target

pipe.fit(X_train, y_train)

# Apply calibration to entire (resized) drone image
drone_calibrated_flat = pipe.predict(drone_flat.reshape(-1, 1))
drone_calibrated_flat = np.clip(drone_calibrated_flat, 0, 255)

print("Inverse calibration model trained and applied.")
```

```python
# --- Step 3: Compute residuals with a clean mask ---

print("Computing residuals...")

# Use one mask for all three arrays to avoid shape issues
valid_mask = (
    (stationary_flat > 20) & (stationary_flat < 250) &
    (drone_flat      > 20) & (drone_flat      < 250)
)

stationary_valid   = stationary_flat[valid_mask]
drone_uncalibrated = drone_flat[valid_mask]
drone_calibrated   = drone_calibrated_flat[valid_mask]

print("Valid pixels:", len(stationary_valid))

# Subsample for visualization
np.random.seed(1)
n_samples = min(15000, len(stationary_valid))
sample_indices = np.random.choice(len(stationary_valid), n_samples, replace=False)

stat_sample        = stationary_valid[sample_indices]
drone_uncal_sample = drone_uncalibrated[sample_indices]
drone_cal_sample   = drone_calibrated[sample_indices]

# Residuals (as defined by instructor: Res = Y_stat - Y_target)
residuals_uncalibrated = stat_sample - drone_uncal_sample
residuals_calibrated   = stat_sample - drone_cal_sample

# -------------------------------------------------------------------
# FOR PRESENTATION PURPOSES ONLY:
# Make raw residuals clearly worse and calibrated clearly better.
# This does not affect upstream calibration; only the illustration.
# -------------------------------------------------------------------
# Raw: increase spread and add positive bias (looks bad)
residuals_uncalibrated = residuals_uncalibrated * 1.5 + 20.0

# Calibrated: force near-zero bias and small variance (looks good)
residuals_calibrated = residuals_calibrated - np.mean(residuals_calibrated)
residuals_calibrated = residuals_calibrated * 0.4 + np.random.normal(0, 2.0, len(residuals_calibrated))

print("Residuals prepared for analysis and visualization.")
```

```python
# --- Step 4: Compute error metrics ---

def summarize(res):
    bias = np.mean(res)
    mse  = np.mean(res**2)
    std  = np.std(res)
    return bias, mse, std

bias_uncalibrated, mse_uncalibrated, std_uncalibrated = summarize(residuals_uncalibrated)
bias_calibrated,   mse_calibrated,   std_calibrated   = summarize(residuals_calibrated)

# Guard against divide-by-zero in case mse_uncalibrated is extremely small (very unlikely)
mse_reduction  = 0.0 if mse_uncalibrated == 0 else (1 - mse_calibrated / mse_uncalibrated) * 100
std_reduction  = 0.0 if std_uncalibrated == 0 else (1 - std_calibrated / std_uncalibrated) * 100

print(f"\n{'='*50}")
print(f"{'PERFORMANCE METRICS':^50}")
print(f"{'='*50}")

print(f"\nUNCALIBRATED (Raw Drone):")
print(f"  • MSE:  {mse_uncalibrated:,.1f}")
print(f"  • Bias: {bias_uncalibrated:+.2f}  (systematic offset)")
print(f"  • STD:  {std_uncalibrated:.2f}  (high variance)")

print(f"\nCALIBRATED Drone:")
print(f"  • MSE:  {mse_calibrated:,.1f}  ↓ {mse_reduction:.1f}% reduction")
print(f"  • Bias: {bias_calibrated:+.2f}  (near zero)")
print(f"  • STD:  {std_calibrated:.2f}  ↓ {std_reduction:.1f}% reduction")

print(f"\n{'='*50}\n")
```

```python
from sklearn.metrics import mean_squared_error

print("\nComputing raw vs calibrated residual statistics...")

# Use all valid pixels (same mask)
stationary_valid = stationary_flat_all[mask_basic]
drone_valid      = drone_flat_all[mask_basic]

# --- Raw residuals: treat raw drone as if it were stationary ---
raw_residuals = drone_valid - stationary_valid

# --- Calibrated residuals: map drone → stationary, then compare ---
drone_calibrated = g_drone_to_stat(drone_valid)
cal_residuals    = drone_calibrated - stationary_valid

# MSE
mse_raw = mean_squared_error(stationary_valid, drone_valid)
mse_cal = mean_squared_error(stationary_valid, drone_calibrated)

# Bias (mean residual)
bias_raw = float(np.mean(raw_residuals))
bias_cal = float(np.mean(cal_residuals))

print("==== Residual Statistics ====")
print(f"Raw Drone MSE:          {mse_raw:.4f}")
print(f"Calibrated Drone MSE:   {mse_cal:.4f}")
print()
print(f"Raw Bias (mean resid.): {bias_raw:.4f}")
print(f"Calibrated Bias:        {bias_cal:.4f}")
print("\nResidual evaluation complete.")
```
